```python

# PHASE 9 - STEP 2: PERFORMANCE MEASUREMENT

from pyspark.sql import functions as F
from time import perf_counter

# Read Bronze table
bronze = spark.table(
    "ecommerce_dev.bronze.bronze_order_items"
)

# Aggregation
analysis_df = (
    bronze
    .groupBy("order_id")
    .agg(
        F.count("*").alias("item_count")
    )
)

# Display execution plan
print("PHYSICAL EXECUTION PLAN")
analysis_df.explain(mode="formatted")

# Measure execution time
start = perf_counter()

result_count = analysis_df.count()

elapsed = perf_counter() - start

print("\nPERFORMANCE RESULTS")
print("Groups returned:", result_count)
print("Elapsed seconds:", round(elapsed, 3))
```

```python

# PHASE 9 - STEP 3: NARROW VS WIDE TRANSFORMATIONS
# Read-only: does not modify any Delta tables

from pyspark.sql import functions as F

df = spark.table(
    "ecommerce_dev.bronze.bronze_order_items"
)

# Narrow transformation: filter rows
narrow_df = df.filter(
    F.col("order_id").isNotNull()
).select("order_id")

print("NARROW TRANSFORMATION PLAN")
narrow_df.explain(mode="formatted")

# Wide transformation: group records by order
wide_df = (
    df.groupBy("order_id")
    .agg(F.count("*").alias("item_count"))
)

print("\nWIDE TRANSFORMATION PLAN")
wide_df.explain(mode="formatted")
```
