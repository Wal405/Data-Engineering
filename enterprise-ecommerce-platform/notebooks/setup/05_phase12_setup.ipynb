```python

from pathlib import Path
import shutil

source = Path(
    "/Workspace/enterprise-ecommerce-platform"
)

destination = Path(
    "/Workspace/Users/"
    "waltahir123_gmail.com#ext#@waltahir123gmail.onmicrosoft.com/"
    "Data-Engineering/enterprise-ecommerce-platform"
)

folders = [
    "config",
    "docs",
    "notebooks",
    "resources",
    "src",
    "tests"
]

def ignore_files(directory, names):
    excluded = {
        ".git",
        ".pytest_cache",
        "__pycache__",
        ".venv",
        ".env",
        ".databricks",
        ".databrickscfg"
    }

    return [
        name for name in names
        if name in excluded
        or name.endswith(".pyc")
        or name.endswith(".csv")
        or name.endswith(".parquet")
        or name.endswith(".json")
    ]

for folder in folders:
    source_folder = source / folder
    destination_folder = destination / folder

    if not source_folder.is_dir():
        print(f"SKIPPED (source missing): {folder}")
        continue

    if destination_folder.exists():
        print(f"SKIPPED (already exists): {folder}")
        continue

    shutil.copytree(
        source_folder,
        destination_folder,
        ignore=ignore_files
    )

    print(f"COPIED: {folder}")

print("\nCopy operation completed.")
```
